```python
def get_validate_data(csv_file, columns):
  _df = pd.read_csv(csv_file)  # ładowanie danych
  df_subset = _df.iloc[:, [2, 4, 5]]  # pobranie wybranych zmiennych
  _df_columns = _df.columns
  if columns.equals(_df_columns):  # inny sposób list(df_columns) == list(df_athens_columns)
    return df_subset
  return None
```
